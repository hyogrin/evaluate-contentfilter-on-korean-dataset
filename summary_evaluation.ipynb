# Simplified Evaluation Summary - Results-Only Markdown Generation

**단순화된 평가 요약** - `results` 폴더의 원본 CSV만 사용하여 마크다운 생성

```python
import sys
import os
sys.path.append(os.path.dirname(os.getcwd()))

from util.md_util import (
    calculate_metrics_from_csv,
    find_results_files,
    generate_content_filter_md,
    generate_content_safety_md,
    generate_performance_comparison_md,
    generate_complete_evaluation_report
)
import pandas as pd
```

## 1. Results 폴더 파일 확인

```python
# Results 폴더의 파일들 확인
results_dir = "results"

print("📁 Results 폴더 파일 목록:")
if os.path.exists(results_dir):
    all_files = os.listdir(results_dir)
    csv_files = [f for f in all_files if f.endswith('.csv')]
    
    print(f"\n총 {len(csv_files)}개 CSV 파일 발견:")
    for f in sorted(csv_files):
        print(f"  - {f}")
        
    print(f"\n기타 파일: {len(all_files) - len(csv_files)}개 (PNG, HTML 등)")
else:
    print("❌ Results 폴더가 없습니다.")
```

## 2. 자동 파일 매핑 테스트

```python
# Content Filter 파일 자동 매핑
print("🔍 Content Filter 파일 자동 매핑:")
cf_files = find_results_files(results_dir, 'content_filter')
for level, path in cf_files.items():
    print(f"  {level}: {os.path.basename(path)}")

# Content Safety 파일 자동 매핑
print("\n🔍 Content Safety 파일 자동 매핑:")
cs_files = find_results_files(results_dir, 'content_safety')
for level, path in cs_files.items():
    print(f"  {level}: {os.path.basename(path)}")

print(f"\n✅ Content Filter: {len(cf_files)}개 레벨 발견")
print(f"✅ Content Safety: {len(cs_files)}개 레벨 발견")
```

## 3. 성능 메트릭 계산 테스트

```python
# 샘플 파일의 메트릭 계산
if cf_files:
    sample_file = list(cf_files.values())[0]
    print(f"📊 샘플 파일 메트릭 계산: {os.path.basename(sample_file)}")
    
    metrics = calculate_metrics_from_csv(f"results/{sample_file}")
    
    print("\n=== 성능 메트릭 ===")
    print(f"Precision: {metrics['precision']:.4f}")
    print(f"Recall: {metrics['recall']:.4f}")
    print(f"F1-Score: {metrics['f1']:.4f}")
    print(f"Accuracy: {metrics['accuracy']:.4f}")
    
    print("\n=== Confusion Matrix ===")
    print(f"TP: {metrics['true_positives']}, TN: {metrics['true_negatives']}")
    print(f"FP: {metrics['false_positives']}, FN: {metrics['false_negatives']}")
    
    print(f"\n총 샘플: {metrics['total_samples']}개")
    print(f"Hate Speech: {metrics['hate_speech_samples']}개")
    print(f"Normal: {metrics['normal_samples']}개")
else:
    print("❌ 테스트할 파일이 없습니다.")
```

## 4. Content Filter 마크다운 생성

```python
# Content Filter 마크다운 테이블 생성
print("📋 Content Filter 마크다운 테이블 생성\n")

cf_md = generate_content_filter_md(results_dir)
print(cf_md)
```

## 5. Content Safety 마크다운 생성

```python
# Content Safety 마크다운 테이블 생성
print("🛡️ Content Safety 마크다운 테이블 생성\n")

cs_md = generate_content_safety_md(results_dir)
print(cs_md)
```

## 6. 성능 비교 테이블

```python
# 성능 비교 테이블 생성
print("⚖️ 성능 비교 테이블 생성\n")

comparison_md = generate_performance_comparison_md(results_dir)
print(comparison_md)
```

## 7. 완전한 평가 보고서 생성

```python
# 전체 보고서 생성
print("📄 완전한 평가 보고서 생성\n")

full_report = generate_complete_evaluation_report(results_dir)
print("=== 완전한 평가 보고서 ===")
print(full_report[:2000])  # 처음 2000자만 출력
if len(full_report) > 2000:
    print("\n... (길이가 길어 생략) ...")
    print(f"\n전체 길이: {len(full_report):,} 문자")
```

## 8. 보고서 파일로 저장

```python
# 최종 보고서를 파일로 저장
output_file = "total_evaluation_report.md"

try:
    full_report = generate_complete_evaluation_report(results_dir)
    
    if full_report.strip():
        with open(output_file, 'w', encoding='utf-8') as f:
            f.write("# Korean Hate Speech Detection Evaluation Report\n\n")
            f.write("## Evaluation Overview \n\n")
            f.write("- **Dataset**: K-MHaS (Korean Multi-label Hate Speech Dataset)\n")
            f.write("- **Evaluation Target**: Azure OpenAI Content Filter vs Azure Content Safety\n")
            f.write("- **Metrics**: Precision, Recall, F1-Score, Accuracy\n")
            f.write("- **Created**: 2025-08-14\n\n")
            f.write(full_report)
            
        file_size = os.path.getsize(output_file)
        print(f"✅ 보고서가 저장되었습니다: {output_file}")
        print(f"📊 파일 크기: {file_size:,} bytes ({file_size/1024:.1f} KB)")
        
    else:
        print("❌ 저장할 보고서 내용이 없습니다.")
        
except Exception as e:
    print(f"❌ 파일 저장 중 오류: {e}")
    import traceback
    traceback.print_exc()
```

## 9. 요약 통계

```python
# 전체 결과 요약
print("📈 최종 요약 통계\n")

# Content Filter 요약
cf_files = find_results_files(results_dir, 'content_filter')
if cf_files:
    print("🔵 Content Filter 요약:")
    cf_metrics = {}
    for level, path in cf_files.items():
        metrics = calculate_metrics_from_csv(path)
        cf_metrics[level] = metrics
        print(f"  {level.capitalize()}: P={metrics['precision']:.3f}, R={metrics['recall']:.3f}, F1={metrics['f1']:.3f}, Acc={metrics['accuracy']:.3f}")

# Content Safety 요약
cs_files = find_results_files(results_dir, 'content_safety')
if cs_files:
    print("\n🟡 Content Safety 요약:")
    cs_metrics = {}
    for level, path in cs_files.items():
        metrics = calculate_metrics_from_csv(path)
        cs_metrics[level] = metrics
        threshold_label = "(1~2)" if level == 'low' else "(3~4)" if level == 'medium' else "(5~6)"
        print(f"  {level.capitalize()} {threshold_label}: P={metrics['precision']:.3f}, R={metrics['recall']:.3f}, F1={metrics['f1']:.3f}, Acc={metrics['accuracy']:.3f}")

# 최고 성능 찾기
if cf_files and cs_files:
    print("\n🏆 최고 성능:")
    
    all_results = []
    
    for level, metrics in cf_metrics.items():
        all_results.append((f"Content Filter ({level})", metrics['f1'], metrics))
    
    for level, metrics in cs_metrics.items():
        threshold_label = "1~2" if level == 'low' else "3~4" if level == 'medium' else "5~6"
        all_results.append((f"Content Safety ({threshold_label})", metrics['f1'], metrics))
    
    # F1-Score 기준으로 정렬
    all_results.sort(key=lambda x: x[1], reverse=True)
    
    print("  F1-Score 순위:")
    for i, (name, f1, metrics) in enumerate(all_results[:3], 1):
        print(f"    {i}. {name}: F1={f1:.3f} (P={metrics['precision']:.3f}, R={metrics['recall']:.3f})")
```
